```python
import matplotlib.pyplot as plt 
import pandas as pd
import seaborn as sns
import plotly.express as tly
import plotly.io as pio
```

```python
tb = pd.read_csv('graficos/dados_barras.csv')
tb


```

```python
fig, ax = plt.subplots(figsize=(8, 5))

sns.barplot(
    data=tb,
    x="categoria",
    y="vendas_medias",
    hue="regiao",
    palette="Set2",
    ax=ax,
)

ax.set_title("Barras", fontsize=14,)
ax.set_xlabel("Categoria", fontsize=11)
ax.set_ylabel("Média de Vendas", fontsize=11)
ax.legend(title="Barras grafico", frameon=True)

plt.tight_layout()
plt.show()
```

```python
tl = pd.read_csv('graficos/dados_linhas.csv')
tl
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.lineplot(
    data=tl,
    x="data",
    y="vendas",
    hue="loja",
    marker="o",
    color="#2b5c8f",
    linewidth=2,
    ax=ax,
)

ax.set_title("Dados de Vendas", fontsize=14, fontweight="bold")
ax.set_xlabel("Datas", fontsize=11)
ax.set_ylabel("Vendas", fontsize=11)

plt.tight_layout()
plt.show()
```
